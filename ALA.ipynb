```python
class Vec:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vec(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Vec({self.x}, {self.y})"

v1 = Vec(1, 2)
v2 = Vec(3, 4)
v3 = v1 + v2
print(f"vector 1:{v1}")
print(f"vector 2:{v2}")
print(f"vector 3:{v3}")
```
